```python
import numpy as np
```

```python
#dataset
x=np.array(
    [
        [0,0],# input features x1 and x2
        [0,1],
        [1,0],
        [1,1]
    ]
)

y=np.array([0,1,1,1])
```

```python
#initlze w,b,n
weights=np.zeros(2)
bias=0
learning_rate=0.01
```

```python
#creating one sing feture from the all  x dataset as nupy doesnot suppoort mutiple data at oNCE
feature=x[0]
target=y[0]
```

```python
#activiation function
def step_funct(z):
    if z >= 0:
        return 1
    else:
        return 0
```

```python
#weighted sum
z=np.dot(feature,weights) + bias

```

```python
#prediction
prediction=step_funct(z)
```

```python
#error
error=target-prediction
```

```python
#fro fisrt sample feature
print(feature)
print(target)
print(z)
print(prediction)
print(error)
```

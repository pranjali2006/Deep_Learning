```python
import numpy as np
```

```python
#dataset
x=np.array(
    [
        [0,0],# input features x1 and x2
        [0,1],
        [1,0],
        [1,1]
    ]
)

y=np.array([0,1,1,1])
```

```python
#initlze w,b,n
weights=np.zeros(2)
bias=0
learning_rate=0.01
```

```python
#creating one sing feture from the all  x dataset as nupy doesnot suppoort mutiple data at oNCE
feature=x[0]
target=y[0]
```

```python
#activiation function
def step_funct(z):
    if z >= 0:
        return 1
    else:
        return 0
```

```python
#weighted sum
z=np.dot(feature,weights) + bias

```

```python
#prediction
prediction=step_funct(z)
```

```python
#error
error=target-prediction
```

```python
#fro fisrt sample feature
print(feature)
print(target)
print(z)
print(prediction)
print(error)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Plot the two classes
plt.scatter(
    x[y == 0, 0], x[y == 0, 1],
    label="Class 0", marker="o"
)

plt.scatter(
    x[y == 1, 0], x[y == 1, 1],
    label="Class 1", marker="x"
)

# Draw the learned decision boundary
x1 = np.linspace(x[:, 0].min() - 1, x[:, 0].max() + 1, 100)

if weights[1] != 0:
    x2 = -(weights[0] * x1 + bias) / weights[1]
    plt.plot(x1, x2, label="Decision boundary")

plt.xlabel("Feature x1")
plt.ylabel("Feature x2")
plt.title("Perceptron Decision Boundary")
plt.legend()
plt.grid(True)
plt.show()
```

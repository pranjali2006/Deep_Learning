gradient descent for a singl artifical neurn using or gate

#steps<br>
1.create datasrt
2.create w,b,n
3.equation of y^
4.calcuatle loss - we will us mse loss function
5.calculate gradient
6.upadet weights and bias
5.repaets teps till efined epochs
6.test the neuron

```python
import numpy as np
```

```python
x=np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
],dtype=float)

y=np.array([0,1,1,1],dtype=float)
```

```python
w=np.zeros((2,1))#2rows and 1 column[[0,.],[0,.]]
b=0.0
n=0.1
epochs=1000
```

```python
for i in range(epochs):

    y_pred=x@w+b #@ performs matrix muttiplication in arrays
    error=y_pred-y.reshape(-1,1)
    loss=np.mean(error**2)#mse=1/n(y-y^)^2  np.mean gives =avg and **2

    gradient_w=(2/len(x))*(x.T @ error)  #taking derrvative of loss function mse
    gradient_b=(2/len(x))*np.sum(error)


    #update the weights and bias
    w -=n*gradient_w
    b -=n*gradient_b

    if i % 100==0:
        print(f"epoch :{epochs},loss :{loss:.4f}")

```
